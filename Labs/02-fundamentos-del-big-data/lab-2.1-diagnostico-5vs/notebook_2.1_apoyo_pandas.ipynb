## 0. Carga del dataset

```python
import pandas as pd
import os

df = pd.read_csv("tienda_online_ventas.csv")
print(f"Dataset cargado: {df.shape[0]} filas x {df.shape[1]} columnas")
df.head()
```

## 1. Volumen

```python
# Tamaño del fichero en disco
tamano_disco_bytes = os.path.getsize("tienda_online_ventas.csv")
tamano_disco_mb = tamano_disco_bytes / (1024 ** 2)
print(f"Tamaño en disco: {tamano_disco_bytes:,} bytes ({tamano_disco_mb:.2f} MB)")

# Tamaño en memoria una vez cargado como DataFrame
tamano_memoria_bytes = df.memory_usage(deep=True).sum()
tamano_memoria_mb = tamano_memoria_bytes / (1024 ** 2)
print(f"Tamaño en memoria (pandas): {tamano_memoria_bytes:,} bytes ({tamano_memoria_mb:.2f} MB)")

print(f"\nFilas: {df.shape[0]:,}")
print(f"Columnas: {df.shape[1]}")
```

```python
# Estimación a escala 50x (si esta tienda tuviera 50 veces más pedidos)
factor = 50
print(f"Tamaño estimado en disco a {factor}x: {tamano_disco_mb * factor:.2f} MB "
      f"(~{tamano_disco_mb * factor / 1024:.3f} GB)")
```

## 3. Veracidad

```python
# Nulos por columna
nulos = df.isnull().sum()
pct_nulos = (nulos / len(df) * 100).round(2)
pd.DataFrame({"nulos": nulos, "pct_nulos": pct_nulos}).sort_values("nulos", ascending=False)
```

```python
# Duplicados: filas exactamente iguales
n_duplicados = df.duplicated().sum()
print(f"Filas duplicadas (idénticas en todas las columnas): {n_duplicados}")
```

```python
# Consistencia de valores en columnas categóricas clave
for columna in ["categoria", "canal", "metodo_pago"]:
    print(f"--- {columna} ---")
    print(df[columna].value_counts())
    print()
```
